```python
import matplotlib.pyplot as plt
from PIL import Image

# Assuming 'img' is your original loaded image from earlier cells
# If 'img' is not defined, please ensure the cell loading it has been run.

# Perform transformations
img_h_flip = img.transpose(Image.FLIP_LEFT_RIGHT)
img_v_flip = img.transpose(Image.FLIP_TOP_BOTTOM)
img_rot_90 = img.transpose(Image.ROTATE_90)
img_rot_270 = img.transpose(Image.ROTATE_270)

# Create a figure with 2 rows and 3 columns for subplots
fig, axes = plt.subplots(2, 3, figsize=(15, 10))

# Display original image
axes[0, 0].imshow(img)
axes[0, 0].set_title('Original Image')
axes[0, 0].axis('off')

# Display horizontal flip
axes[0, 1].imshow(img_h_flip)
axes[0, 1].set_title('Horizontal Flip')
axes[0, 1].axis('off')

# Display vertical flip
axes[0, 2].imshow(img_v_flip)
axes[0, 2].set_title('Vertical Flip')
axes[0, 2].axis('off')

# Display 90-degree rotation
axes[1, 0].imshow(img_rot_90)
axes[1, 0].set_title('90-degree Rotation')
axes[1, 0].axis('off')

# Display 270-degree rotation
axes[1, 1].imshow(img_rot_270)
axes[1, 1].set_title('270-degree Rotation')
axes[1, 1].axis('off')

# Hide the unused subplot
axes[1, 2].axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np

# Ensure 'img' (PIL Image object) is available from previous cells.
# Convert PIL Image to NumPy array (RGB format)
img_np_rgb = np.array(img)

# Convert RGB to BGR for OpenCV processing
img_cv_bgr = cv2.cvtColor(img_np_rgb, cv2.COLOR_RGB2BGR)
```

```python
import matplotlib.pyplot as plt

# Display the original RGB NumPy array
plt.figure(figsize=(6, 6))
plt.imshow(img_cv_bgr )
plt.title('Original Image (NumPy RGB)')
plt.axis('off')
plt.show()
```

```python
import matplotlib.pyplot as plt
import cv2

# Display the BGR converted image (converted back to RGB for display)
plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(img_cv_bgr, cv2.COLOR_BGR2RGB))
plt.title('Converted to BGR (Displayed as RGB)')
plt.axis('off')
plt.show()
```

```python
# Horizontal Flip using OpenCV
h_flip_img_bgr = cv2.flip(img_cv_bgr, 1) # 1 for horizontal flip
h_flip_img_rgb = cv2.cvtColor(h_flip_img_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(h_flip_img_rgb)
plt.title('Horizontal Flip (OpenCV)')
plt.axis('off')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Convert BGR back to RGB using OpenCV and display
img_cv_rgb = cv2.cvtColor(img_cv_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(img_cv_rgb)
plt.title('Image after BGR to RGB Conversion')
plt.axis('off')
plt.show()
```

```python
# Vertical Flip using OpenCV
v_flip_img_bgr = cv2.flip(img_cv_bgr, 0) # 0 for vertical flip
v_flip_img_rgb = cv2.cvtColor(v_flip_img_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(v_flip_img_rgb)
plt.title('Vertical Flip (OpenCV)')
plt.axis('off')
plt.show()
```

```python
# Create a figure for rotations with subplots
fig_rot, axes_rot = plt.subplots(1, 2, figsize=(12, 6))

# 90-degree Rotation (Clockwise) using OpenCV
rot_90_img_bgr = cv2.rotate(img_cv_bgr, cv2.ROTATE_90_CLOCKWISE)
rot_90_img_rgb = cv2.cvtColor(rot_90_img_bgr, cv2.COLOR_BGR2RGB)

axes_rot[0].imshow(rot_90_img_rgb)
axes_rot[0].set_title('90-degree Rotation (OpenCV)')
axes_rot[0].axis('off')

# 270-degree Rotation (Clockwise) using OpenCV, equivalent to 90-degree counter-clockwise
rot_270_img_bgr = cv2.rotate(img_cv_bgr, cv2.ROTATE_90_COUNTERCLOCKWISE)
rot_270_img_rgb = cv2.cvtColor(rot_270_img_bgr, cv2.COLOR_BGR2RGB)

axes_rot[1].imshow(rot_270_img_rgb)
axes_rot[1].set_title('270-degree Rotation (OpenCV)')
axes_rot[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Load the CSV file into a DataFrame
df_marks = pd.read_csv('/content/Simple 100 Student Marks.csv')

# Display the first few rows of the DataFrame to understand its structure
print('First 5 rows of the DataFrame:')
display(df_marks.head())
```

```python
# Display basic information about the DataFrame, including data types and non-null values
print('\nDataFrame Info:')
df_marks.info()
```

### Statistical Operations

```python
# Calculate descriptive statistics for all numerical columns
print('Descriptive Statistics for all numerical columns:')
display(df_marks.describe())
```

```python
# Calculate mean, median, and mode for the 'maths_marks' column
mean_maths_marks = df_marks['maths_marks'].mean()
median_maths_marks = df_marks['english_marks'].median()
mode_maths_marks = df_marks['science_marks'].mode().tolist() # mode can return multiple values

print(f"Mean Maths Marks: {mean_maths_marks:.2f}")
print(f"Median Maths Marks: {median_maths_marks}")
print(f"Mode Maths Marks: {mode_maths_marks}")
```

```python
# Find the minimum and maximum 'maths_marks'
min_maths_marks = df_marks['maths_marks'].min()
max_maths_marks = df_marks['english_marks'].max()

print(f"Minimum Maths Marks: {min_maths_marks}")
print(f"Maximum Maths Marks: {max_maths_marks}")
```

### Comparison Operations

```python
# Filter students who scored more than 70 in 'maths_marks'
students_above_70_maths = df_marks[df_marks['maths_marks'] > 70]
print('Students with Maths Marks > 70:')
display(students_above_70_maths)
```

```python
# Compare students based on their performance in two subjects: 'maths_marks' vs 'science_marks'
# Find students who scored higher in Maths than Science
students_maths_higher_than_science = df_marks[df_marks['maths_marks'] > df_marks['science_marks']]
print('Students who scored higher in Maths than Science:')
display(students_maths_higher_than_science)
```

```python
# Find students with 'Excellent' performance, defined as 'maths_marks' >= 90
students_excellent_maths = df_marks[df_marks['maths_marks'] >= 90]
print('Students with Excellent Maths Marks (>=90):')
display(students_excellent_maths)
```
