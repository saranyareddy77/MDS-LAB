```python
# Create a NumPy array of strings
string_array = np.array(['apple', 'banana', 'cherry', 'date'])
print("Original string array:", string_array)
print("Data type of string array:", string_array.dtype)
```

```python
# Accessing elements
print("First element:", string_array[0])
print("Slice of elements:", string_array[1:3])

# Element-wise string operations using np.char
# np.char provides a set of vectorized string operations.

# Convert to uppercase
uppercase_array = np.char.upper(string_array)
print("Uppercase array:", uppercase_array)

# Concatenate strings
concatenated_string = np.char.add(string_array, '_fruit')
print("Concatenated array:", concatenated_string)

# Check if a substring exists
contains_e = np.char.find(string_array, 'e') != -1
print("Contains 'e' array:", contains_e)

# Replace a substring
replaced_array = np.char.replace(string_array, 'a', 'x')
print("Replaced 'a' with 'x':", replaced_array)
```
