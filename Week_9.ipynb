```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('/content/hospital_patients_real_world.csv')
display(df.head())
```

```python
print("Missing values before filling:")
print(df.isnull().sum())
```

```python
# Convert blank strings to NaN
df = df.replace(r'^\s*$', pd.NA, regex=True)

print("\nMissing values after detecting blanks:")
print(df.isna().sum())
```

```python
# Handle missing values
if df["Age"].isna().sum() > 0:
    df["Age"] = df["Age"].fillna(df["Age"].median())

if df["Gender"].isna().sum() > 0:
    df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])

if df["Diagnosis"].isna().sum() > 0:
    df["Diagnosis"] = df["Diagnosis"].fillna(df["Diagnosis"].mode()[0])
```

```python
# Check again
print("\nMissing values AFTER handling:")
print(df.isna().sum())
```

```python
# Fill numerical column with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill categorical columns with mode
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])
df["Diagnosis"] = df["Diagnosis"].fillna(df["Diagnosis"].mode()[0])
```

```python
print("Missing values after filling:")
print(df.isnull().sum())
```

```python
# 4. Detect outliers using IQR method
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("\nQ1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
# Detect outliers
outliers = df[
    (df["Age"] < lower_limit) |
    (df["Age"] > upper_limit)
]

print("\nNumber of outliers:", len(outliers))
print("Outliers:")
print(outliers)
```

```python
# 5. Filter out outliers
df_filtered = df[
    (df["Age"] >= lower_limit) &
    (df["Age"] <= upper_limit)
]

print("\nData after removing outliers:")
print(df_filtered.head())

print("\nFinal DataFrame Shape:", df_filtered.shape)
```

```python
import pandas as pd # Import the pandas library for data manipulation

# Load the 'student_performance_large.csv' dataset into a DataFrame
df = pd.read_csv("student_performance_large.csv")

# Print a header and display the first 5 rows of the original dataset
print("Original Dataset:")
print(df.head())
```

```python
print("Missing values BEFORE handling:")
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["StudyHours"] = df["StudyHours"].fillna(df["StudyHours"].median())
df["Marks"] = df["Marks"].fillna(df["Marks"].median())
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].median())
```

```python
print("Missing values AFTER handling:")
print(df.isnull().sum())
```

```python
# Function to detect outliers
def detect_outliers(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower) | (df[column] > upper)]

    print("\nColumn:", column)
    print("Q1 =", Q1)
    print("Q3 =", Q3)
    print("IQR =", IQR)
    print("Lower Bound =", lower)
    print("Upper Bound =", upper)
    print("Outliers:")
    print(outliers[[column]])

# Detect outliers
detect_outliers("StudyHours")
detect_outliers("Marks")
```

```python
# Calculate IQR for StudyHours
Q1 = df["StudyHours"].quantile(0.25)
Q3 = df["StudyHours"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df_filtered = df[
    (df["StudyHours"] >= lower) &
    (df["StudyHours"] <= upper)
]

# Calculate IQR for Marks
Q1 = df_filtered["Marks"].quantile(0.25)
Q3 = df_filtered["Marks"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df_filtered = df_filtered[
    (df_filtered["Marks"] >= lower) &
    (df_filtered["Marks"] <= upper)
]

print("Dataset after removing outliers:")
print(df_filtered)

print("\nFinal shape:")
print(df_filtered.shape)
```
