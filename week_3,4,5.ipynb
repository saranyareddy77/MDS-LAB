### Deleting Records (Rows) Based on a Condition

To delete rows that meet a specific condition, you can use boolean indexing. For example, let's delete all artists whose 'Sex' is 'Male' to demonstrate.

```python
# Example: Delete rows where 'Sex' is 'Male'
original_rows = df.shape[0]
df_filtered_rows = df[df['Sex'] != 'Male'].copy()
print(f"Original number of rows: {original_rows}")
print(f"Number of rows after deleting 'Male' artists: {df_filtered_rows.shape[0]}")
display(df_filtered_rows.head())
```

### Deleting Columns

To delete columns, you can use the `drop()` method. Let's demonstrate by dropping the 'Debut Year' column.

```python
# Example: Delete the 'Debut Year' column
original_columns = df.shape[1]
df_filtered_columns = df.drop(columns=['Debut Year']).copy()
print(f"Original number of columns: {original_columns}")
print(f"Number of columns after deleting 'Debut Year': {df_filtered_columns.shape[1]}")
display(df_filtered_columns.head())
```

### Sorting Records

Let's sort the DataFrame by 'Total Streams (in millions)' in descending order to see the artists with the highest streams at the top. We'll create a new DataFrame `df_sorted` for this.

```python
# Sort the DataFrame by 'Total Streams (in millions)' in descending order
df_sorted = df.sort_values(by='Total Streams (in millions)', ascending=False).copy()

print("DataFrame sorted by Total Streams (top 5):")
display(df_sorted.head())
```

### Ranking Records

Now, let's add a new column named 'Stream Rank' to the DataFrame, ranking artists based on their 'Total Streams (in millions)'. We'll use the `df_sorted` DataFrame for this to apply rank on already sorted data.

```python
# Add a 'Stream Rank' column based on 'Total Streams (in millions)'
df_ranked = df_sorted.copy()
df_ranked['Stream Rank'] = df_ranked['Total Streams (in millions)'].rank(ascending=False, method='min')

print("DataFrame with Stream Rank (top 5):")
display(df_ranked.head())
```

### Value Counts for a Categorical Column

Now, let's examine the distribution of a categorical column, for example, 'Primary Genre', using `value_counts()`.

```python
# Display value counts for 'Primary Genre'
print("Value counts for 'Primary Genre':")
display(df['Primary Genre'].value_counts())
```

### Mean, Median, and Mode for a Specific Numerical Column

Let's calculate the mean, median, and mode for 'Total Streams (in millions)' to understand its central tendency.

```python
# Calculate mean, median, and mode for 'Total Streams (in millions)'
mean_streams = df['Total Streams (in millions)'].mean()
median_streams = df['Total Streams (in millions)'].median()
mode_streams = df['Total Streams (in millions)'].mode()

print(f"Mean of Total Streams (in millions): {mean_streams:.2f}")
print(f"Median of Total Streams (in millions): {median_streams:.2f}")
print(f"Mode of Total Streams (in millions): {mode_streams.tolist()}")
```

### Renaming a Single Column

To rename a single column, we can use the `.rename()` method with a dictionary mapping the old column name to the new column name. Let's rename 'Artist Name' to 'Artist'.

```python
# Rename a single column: 'Artist Name' to 'Artist'
df_single_renamed = df.rename(columns={'Artist Name': 'Artist'}).copy()

print("DataFrame with 'Artist Name' renamed to 'Artist' (first 5 rows):")
display(df_single_renamed.head())
```

### Renaming Multiple Columns

We can also rename multiple columns at once by providing a dictionary with all the old-to-new column name mappings to the `.rename()` method. Let's rename 'Sex' to 'Gender' and 'Primary Language' to 'Language'.

```python
# Rename multiple columns: 'Sex' to 'Gender' and 'Primary Language' to 'Language'
df_multi_renamed = df.rename(columns={'Sex': 'Gender', 'Primary Language': 'Language'}).copy()

print("DataFrame with 'Sex' renamed to 'Gender' and 'Primary Language' to 'Language' (first 5 rows):")
display(df_multi_renamed.head())
```

## Creating Pandas Series and DataFrames

Let's explore how to create Pandas Series and DataFrames from various Python data structures.

### Creating a Series from a List

A Pandas Series can be created from a simple Python list.

```python
# Create a Series from a list
data_list = [10, 20, 30, 40, 50]
series_from_list = pd.Series(data_list)

print("Series created from a list:")
display(series_from_list)
```

### Creating a Series from a Dictionary

A Series can also be created from a Python dictionary, where keys become the index and values become the data.

```python
# Create a Series from a dictionary
data_dict = {'a': 100, 'b': 200, 'c': 300}
series_from_dict = pd.Series(data_dict)

print("Series created from a dictionary:")
display(series_from_dict)
```

### Creating a DataFrame from a Dictionary of Lists

A common way to create a DataFrame is from a dictionary where each key represents a column name and its value is a list of data for that column.

```python
# Create a DataFrame from a dictionary of lists
data_df_dict = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
}
df_from_dict_of_lists = pd.DataFrame(data_df_dict)

print("DataFrame created from a dictionary of lists:")
display(df_from_dict_of_lists)
```

### Creating a DataFrame from a List of Dictionaries

Another way to create a DataFrame is from a list where each dictionary in the list represents a row, and the keys are column names.

```python
# Create a DataFrame from a list of dictionaries
data_df_list = [
    {'Name': 'Eve', 'Age': 22, 'City': 'Miami'},
    {'Name': 'Frank', 'Age': 40, 'City': 'Boston'},
    {'Name': 'Grace', 'Age': 33, 'City': 'Seattle'}
]
df_from_list_of_dicts = pd.DataFrame(data_df_list)

print("DataFrame created from a list of dictionaries:")
display(df_from_list_of_dicts)
```
